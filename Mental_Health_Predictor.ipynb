```python
#data cleaning
df_clean= df.copy()
print("Working copy created. Shape: ", df_clean.shape)
df_clean['Age'].min()
```

```python
#fixing age
print("Before cleaning: ")
print(f"Rows: {len(df_clean)}")
print(f"Age Range:{df_clean['Age'].min()} to {df_clean['Age'].max()}")
```

```python
#fixing gender
#first standardize, strip spaces and convert to lower cases
df_clean['Gender']=df_clean['Gender'].str.strip().str.lower()

#grouping
male_terms=['Male','male','m','M','Make','Male', 'Cis Male','Man','Male-ish','Male (CIS)','Mal','something kinda male? ','maile ','Guy (ish) ^_^ ',
'msle ','Mail','cis male','Malr','Cis Man','ostensibly male, unsure what that really means','male leaning androgynous']
female_terms=['Female','female','f','F','Woman','Female','Female (trans)','Trans-female','Cis Female','Femake','Trans woman','cis-female/femme','Female (cis)','femail']
```

```python
def clean_gender(g):
  g= str(g).strip().lower()
  if g in male_terms:
    return 'Male'
  elif g in female_terms:
    return 'Female'
  else:
    return 'Other'

df_clean['Gender']=df_clean['Gender'].apply(clean_gender)
print("Gender Distribution after Cleaning: ")
print(df_clean['Gender'].value_counts())
print(f"Unique Values now: {df_clean['Gender'].nunique()}")
```

```python
#visualizing gender distribution after cleaning
plt.figure(figsize=(7,4))
sns.countplot(data=df_clean, x='Gender', hue="treatment", palette={'Yes':'#2ecc71', 'No':'#e74c3c'})
plt.title("Treatment Sought By Gender")
plt.xlabel("Gender")
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
#fix missing values
print("Missing values before fix: ")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
df_clean['work_interfere']=df_clean['work_interfere'].fillna("Never")
#'work_interfere' is the main column with most NaNs
#these are people who didnt answer the survey most likely becos mental health does not interfere with their work
# so chose 'Never' to replace the NaNs
df_clean['self_employed']=df_clean['self_employed'].fillna(df_clean['self_employed'].mode()[0])
#'self_employed' NaNs filled with most common value 'mode'
```

```python
print("Missing values after fix: ")
missing=df_clean.isnull().sum()[df_clean.isnull().sum()>0]
print(missing if len(missing)> 0 else "None - all clean")
```

```python
#drop columns that arent needed
cols_to_drop=['Timestamp','Country','state','comments']
df_clean=df_clean.drop(columns=cols_to_drop, errors='ignore')
print("Remaining columns: ", df_clean.shape[1])
print(df_clean.columns.to_list())
```

```python
#label encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
#identifying all categorical columns(non-numeric columns)
categ_cols=df_clean.select_dtypes(include=['object']).columns.to_list()
print("Columns to Encode: ", categ_cols)
```

```python
for col in categ_cols:
  df_clean[col]=le.fit_transform(df_clean[col].astype(str))

print("All columns are now numeric: \n", df_clean.dtypes)
```

```python
print("First the rows after encoding: ")
df_clean.head(3)
```

```python
#train_test_split
from sklearn.model_selection import train_test_split

x=df_clean.drop('treatment',axis=1)
y=df_clean['treatment']

print("Features shape: ", x.shape)
print("Target shape: ", y.shape)

x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(f"Training set:{x_train.shape[0]} rows ")
print(f"Testing set: {x_test.shape[0]} rows")
```

```python
#importing models
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix)
from sklearn.preprocessing import StandardScaler
print("All Libraries Imported....")
```

```python
scaler= StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)


print(f"x_train_scaled shape: {x_train_scaled.shape}")
print(f"x_test_scaled shape: {x_test_scaled.shape}")
```

```python
#training
rf_model=RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
rf_model.fit(x_train_scaled, y_train)
```

```python
#prediction
rf_pred=rf_model.predict(x_test_scaled)
print("Prediction Complete..")
```

```python
print(f"First 10 predictions:  {rf_pred[:10]}")
print(f"\n First 10 actuals:  {y_test.values[:10]}")
print("\n 0 = No Treatment")
print("\n 1 = Sought Treatment")
```

```python
#evaluation
ac=accuracy_score(y_test, rf_pred)
cm=confusion_matrix(y_test, rf_pred)
cr=classification_report(y_test,rf_pred)
```

```python
print(f"Accuracy Score= {ac}")
print("="*40)
print(f"\n Confusion Matrix:\n {cm}\n")
print("="*40)
print(f"\n Classification Report: \n{cr}")
```

```python
from matplotlib.colors  import ListedColormap
my_colors=["red","green",]
custom_cmap=ListedColormap(my_colors)
```

```python
sns.heatmap(cm,annot=True,fmt='d',cmap=custom_cmap)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Value")
plt.ylabel("Actual Value")
plt.tight_layout()
plt.savefig("cm.png",dpi=200,bbox_inches='tight')

plt.show()
```
